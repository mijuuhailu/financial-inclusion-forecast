Insight 1

The dataset is primarily focused on the Access and Usage pillars, which aligns with the project's objective of forecasting account ownership and digital payment adoption.

Insight 2

Most records come from operator reports and nationally representative surveys, providing complementary supply-side and demand-side perspectives.

Insight 3

Approximately 93% of records have high confidence, indicating that the dataset is built largely from authoritative sources.

Insight 4

Indicator coverage is uneven. Core indicators such as account ownership have multiple observations, while many operational indicators (e.g., M-Pesa users and Telebirr transaction values) have only one observation. These sparse indicators will require enrichment before they can effectively support forecasting.

Insight 5

The event timeline captures major milestones in Ethiopia's digital finance ecosystem, including the launch of Telebirr, the entry of M-Pesa, the rollout of Fayda Digital ID, and payment infrastructure developments. These events provide a strong basis for later event-impact modeling.

```python
import pandas as pd

# ----------------------------
# Internet Penetration Records
# Source: World Bank (IT.NET.USER.ZS)
# ----------------------------

internet_data = [
    ("2020", 16.1),
    ("2021", 16.7),
    ("2022", 18.0),
    ("2023", 20.3),
    ("2024", 21.9)
]

# Find the next available record number
last_id = (
    df["record_id"]
    .str.extract(r"REC_(\d+)")
    .astype(float)
    .max()[0]
)

next_id = int(last_id) + 1

new_records = []

for year, value in internet_data:

    new_records.append({

        "record_id": f"REC_{next_id:04d}",
        "record_type": "observation",
        "category": None,

        "pillar": "USAGE",

        "indicator": "Internet Penetration",

        "indicator_code": "USG_INTERNET_PEN",

        "indicator_direction": "higher_better",

        "value_numeric": value,

        "value_text": None,

        "value_type": "percentage",

        "unit": "%",

        "observation_date": f"{year}-12-31",

        "period_start": None,

        "period_end": None,

        "fiscal_year": year,

        "gender": "all",

        "location": "Ethiopia",

        "region": None,

        "source_name": "World Bank",

        "source_type": "research",

        "source_url": "https://data.worldbank.org/indicator/IT.NET.USER.ZS?locations=ET",

        "confidence": "high",

        "related_indicator": None,

        "relationship_type": None,

        "impact_direction": None,

        "impact_magnitude": None,

        "impact_estimate": None,

        "lag_months": None,

        "evidence_basis": None,

        "comparable_country": "",

        "collected_by": "mijuu",

        "collection_date": pd.Timestamp.today().strftime("%Y-%m-%d"),

        "original_text": f"Individuals using the Internet (% of population): {value}",

        "notes": (
            "Added during Task 1 enrichment. "
            "Internet penetration is a leading indicator of digital payment adoption."
        )

    })

    next_id += 1

# Convert to dataframe
internet_df = pd.DataFrame(new_records)

# Append
df = pd.concat([df, internet_df], ignore_index=True)

# Ensure datetime
df["observation_date"] = pd.to_datetime(df["observation_date"])

# Verify
print(df[df["indicator_code"]=="USG_INTERNET_PEN"])

print(f"\nDataset shape: {df.shape}")
```
